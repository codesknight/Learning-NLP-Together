```python
import requests
from bs4 import BeautifulSoup

# 目标URL
url = 'https://www.bejson.com/rarechar/'

# 发出HTTP请求获取网页内容
response = requests.get(url)

con_list = []

# 检查请求是否成功
if response.status_code == 200:
    # 手动设置编码
    response.encoding = response.apparent_encoding
    # 使用BeautifulSoup解析HTML内容
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # 查找所有<a>标签中的<span>标签
    anchors = soup.find_all('a')  # 查找所有a标签
    
    for index, anchor in enumerate(anchors, 1):
        # 在每个<a>标签中查找<span>标签
        spans = anchor.find_all('span')
        for span in spans:
            print(f"Anchor {index} Span: {span.get_text()}")
            con_list.append(span.get_text())
else:
    print(f"Failed to retrieve the webpage. Status code: {response.status_code}")
    
len(con_list)
```

```python
# 读取txt中的生僻字，存储在列表中
spz_list = []
#with open('生僻字3.txt', 'r', encoding='utf-8') as file:
with open('spz.txt', 'r', encoding='utf-8') as file:    
    lines = file.readlines()
    i = 0
    for line in lines:
        print(line.strip())  # 去掉末尾换行符
        spz_list.append(line.strip())
        i= i+1
    print(spz_list)
    print(i/2)# 6318

# 使用字典推导式实现转换
spz_dict = {spz_list[i+1]: spz_list[i] for i in range(0, len(spz_list) - 1, 2)}

print(spz_dict)
len(spz_dict)# 6318
```
